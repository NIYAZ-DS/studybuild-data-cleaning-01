```python
int_types = ['age', 'purchase_count', 'last_purchase_days', 'returned_items', 'satisfaction_score']
cat_types = ['customer_id', 'gender', 'city', 'province', 'membership_tier', 'payment_method', 'device', 'discount_used']

for column in ds.columns:
    if column in int_types:
        ds[column] = ds[column].astype('int')

    elif column in cat_types :
        ds[column] = ds[column].astype('category')

    elif column == 'signup_date':
        ds[column] = pd.to_datetime(ds[column])
```
