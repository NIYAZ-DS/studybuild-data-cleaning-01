```python
# checking outliers using the IQR method
numeric_cols = ["age","purchase_count","avg_order_value","total_spending","last_purchase_days","returned_items","satisfaction_score"]

for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    
    print(col, ":", len(outliers), "outliers")
```

```python
# visualizing possible outliers
for col in numeric_cols:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

```python
# creating a copy of the original dataset for cleaning
df_clean = df.copy()
```

```python
# removing exact duplicate rows
df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))
```

```python
# replacing invalid ages with missing values
df_clean.loc[(df_clean["age"] < 18) | (df_clean["age"] > 100),"age"] = np.nan
```

```python
# filling missing age values with the median
age_median = df_clean["age"].median()
print("Age median:", age_median)

df_clean["age"] = df_clean["age"].fillna(age_median)
```

```python
# fixing missing and inconsistent total_spending values
expected = df_clean["avg_order_value"] * df_clean["purchase_count"]
diff = (df_clean["total_spending"] - expected).abs()

print("Inconsistent values:", (diff > 1).sum())

df_clean.loc[diff > 1,"total_spending"] = expected[diff > 1]

mask = df_clean["total_spending"].isna()
df_clean.loc[mask,"total_spending"] = expected[mask]
```

```python
# creating an inferred gender column from first_name
name_gender = {
    "Ali":"M",
    "Amir":"M",
    "Arash":"M",
    "Reza":"M",
    "Sina":"M",
    "Parsa":"M",
    "Kimia":"F",
    "Maryam":"F",
    "Mina":"F",
    "Neda":"F",
    "Sara":"F",
    "Zahra":"F"
}

df_clean["gender_inferred"] = df_clean["first_name"].map(name_gender)

df_clean[["customer_id","first_name","gender","gender_inferred"]].head(15)
```

```python
# checking the zero purchase anomaly
df_clean[df_clean["customer_id"] == 1024][["customer_id","avg_order_value","purchase_count","total_spending","returned_items"]]
```

```python
# fixing purchase_count for customer 1024
df_clean.loc[df_clean["customer_id"] == 1024,"purchase_count"] = np.nan

purchase_median = df_clean.loc[df_clean["purchase_count"] > 0,"purchase_count"].median()
print("Median purchase_count:", purchase_median)

df_clean["purchase_count"] = df_clean["purchase_count"].fillna(purchase_median)

df_clean.loc[df_clean["customer_id"] == 1024,"total_spending"] = (
    df_clean.loc[df_clean["customer_id"] == 1024,"avg_order_value"] *
    df_clean.loc[df_clean["customer_id"] == 1024,"purchase_count"]
)

df_clean[df_clean["customer_id"] == 1024]
```

```python
# converting integer columns to int
df_clean["age"] = df_clean["age"].astype(int)
df_clean["purchase_count"] = df_clean["purchase_count"].astype(int)

df_clean[["age","purchase_count"]].dtypes
```

```python
# removing extra spaces from text columns
text_cols = df_clean.select_dtypes(include="object").columns

for col in text_cols:
    df_clean[col] = df_clean[col].str.strip()
```

```python
# making sure signup_date is stored as datetime
df_clean["signup_date"] = pd.to_datetime(df_clean["signup_date"], errors="coerce")
```

```python
# resetting the dataframe index
df_clean.reset_index(drop=True, inplace=True)
```

```python
# checking missing values after cleaning
print(df_clean.isnull().sum())
print("Total missing values:", df_clean.isnull().sum().sum())
```

## Cleaning Summary

- Removed the exact duplicated row for customer_id 1014.
- Replaced the invalid age value of 145 with a missing value and filled it using the median age.
- Converted signup_date from object to datetime.
- Detected the inconsistent total_spending value of customer_id 1030 and reconstructed it using purchase_count * avg_order_value.
- Reconstructed the missing total_spending value for customer_id 1040 using the same formula.
- Investigated the inconsistency between first_name and gender and added a separate gender_inferred column without changing the original gender column.
- Investigated customer_id 1024, where purchase_count was 0 despite having returned_items. The purchase_count was replaced using the median of valid non-zero purchase counts and total_spending was recalculated.
- Converted age and purchase_count to integer data types after imputation.
- Checked categorical values, date validity, extra spaces, numeric ranges, and outliers.
- Kept the remaining returned_items > purchase_count cases unchanged because the available data was not enough to prove that they were incorrect.
- Validated the cleaned dataset after all changes.
